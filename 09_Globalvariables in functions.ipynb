```python
student = "Anu"
marks = 75

def display():
    print("Student:", student)
    print("Marks:", marks)

def update():
    global marks
    marks = int(input("Enter new marks: "))
    print("Marks updated.")

def grade():
    if marks >= 90:
        print("Grade: A+")
    elif marks >= 75:
        print("Grade: A")
    elif marks >= 60:
        print("Grade: B")
    else:
        print("Grade: C")

display()
update()
display()
grade()
```
