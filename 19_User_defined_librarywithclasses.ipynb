```python
%%writefile student_library.py

class Student:

    def __init__(self):
        self.students = {}

    def add_student(self, roll, name):
        self.students[roll] = name

    def display(self):
        for roll, name in self.students.items():
            print(roll, "-", name)

    def search(self, roll):
        if roll in self.students:
            print("Student:", self.students[roll])
        else:
            print("Student not found.")

    def update(self, roll, name):
        if roll in self.students:
            self.students[roll] = name
            print("Student updated.")
        else:
            print("Student not found.")
```

```python
from student_library import Student

s = Student()

s.add_student(101, "Anu")
s.add_student(102, "Meera")

print("Student Details:")
s.display()

print("\nSearching Student:")
s.search(101)

print("\nUpdating Student:")
s.update(102, "Meera Thomas")

print("\nUpdated Details:")
s.display()
```
