```python
buffer_size = 10

print("Maximum buffer size:", buffer_size)

data = input("Enter data: ")

print("\nInput length:", len(data))

if len(data) > buffer_size:
    print("⚠ Buffer Overflow Detected!")
    print("Allowed:", buffer_size)
    print("Entered:", len(data))
else:
    print("Data stored successfully.")
    print("No overflow detected.")
```
