```python
import random

limit = 5
requests = 10

print("SYN Flood Simulation")
print("Normal connection limit:", limit)

for i in range(1, requests + 1):

    ip = "192.168.1." + str(random.randint(1, 254))

    if i <= limit:
        print("SYN request from", ip, "- Accepted")
    else:
        print("SYN request from", ip, "- Rejected")

print("\nSimulation completed.")
```
