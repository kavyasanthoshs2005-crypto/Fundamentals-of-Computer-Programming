## Object-Oriented Programming
### Fundamentals of Computer Programming

##  What is Object-Oriented Programming?

Object-Oriented Programming (OOP) is a programming method
where we organize a program using **objects and classes**.

### Main Concepts:
- Class
- Object
- Attributes
- Methods

##  What is a Class?

A **class** is a blueprint or design used to create objects.

For example:
Student is a class.

It can contain:
    
• Name
    
• Age

• Introduction

### A class can contain:
- Attributes → information about an object
- Methods → actions performed by an object

##  What is an Object?

An **object** is a real instance created from a class.

For example:

Student → Class

student1 → Object

The object contains actual values such as:

Name = Kavya

Age = 21

#  Example Program – Creating a Class and Object

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print("My name is", self.name)
        print("My age is", self.age)


student1 = Student("Kavya", 21)

student1.introduce()
```

##  Explanation

### class Student:
Creates a class called Student.

### __init__(self, name, age):
Constructor that runs automatically when an object is created.

### self.name = name
Stores the student's name.

### self.age = age
Stores the student's age.

### introduce()
A method that displays the student's details.

### student1 = Student("Kavya", 21)
Creates an object called student1.

### student1.introduce()
Calls the introduce() method using the object.

##  Extra Step – Creating Multiple Objects

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print("My name is", self.name)
        print("My age is", self.age)


student1 = Student("Kavya", 21)
student2 = Student("Manju Sree", 22)

student1.introduce()
student2.introduce()
```

##  Explanation of the Extra Step

In the first program, we created only one object:

student1 = Student("Kavya", 21)

Now, we create another object:

student2 = Student("Manju Sree", 22)

Both objects are created from the same **Student class**.

But each object has different data.

student1 → Kavya, 21

student2 → Manju Sree, 22

This shows that **one class can create multiple objects**.

##  Advantages of Classes and Objects

1. Code can be reused.
2. Programs become easier to organize.
3. Data and functions can be grouped together.
4. Large programs become easier to manage.
5. Multiple objects can be created from one class.

#  Conclusion

A **class** is a blueprint used to create objects.

An **object** is an actual instance of a class.

In our example:

Student → Class
    
student1 → Object
    
student2 → Object

Therefore, one class can be used to create multiple objects
with different data.
