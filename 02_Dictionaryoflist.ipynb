```python
students = {
    "Cyber Security": ["Anu", "Meera"],
    "Computer Science": ["Rahul", "Arun"]
}

while True:
    print("\n1. Add Student")
    print("2. Remove Student")
    print("3. Search Student")
    print("4. Display Department")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        dept = input("Enter department: ")
        name = input("Enter student name: ")

        if dept not in students:
            students[dept] = []

        students[dept].append(name)
        print("Student added.")

    elif choice == "2":
        name = input("Enter name to remove: ")

        for dept in students:
            if name in students[dept]:
                students[dept].remove(name)
                print("Student removed.")
                break
        else:
            print("Student not found.")

    elif choice == "3":
        name = input("Enter name to search: ")

        for dept in students:
            if name in students[dept]:
                print(name, "belongs to", dept)
                break
        else:
            print("Student not found.")

    elif choice == "4":
        dept = input("Enter department: ")
        print(students.get(dept, "Department not found"))

    elif choice == "5":
        break
```
