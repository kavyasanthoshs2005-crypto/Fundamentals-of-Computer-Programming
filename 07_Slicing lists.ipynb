```python

movies = [
    "Leo",
    "Jailer",
    "Vikram",
    "Master",
    "Kaithi",
    "Beast",
    "Doctor",
    "Coolie"
]

print("My Movie Playlist:")
print(movies)

print("\nFirst 3 Movies:")
print(movies[:3])

print("\nMovies from 3rd to 6th:")
print(movies[2:6])

print("\nLast 3 Movies:")
print(movies[-3:])

print("\nAlternate Movies:")
print(movies[::2])

print("\nReverse Playlist:")
print(movies[::-1])
```
