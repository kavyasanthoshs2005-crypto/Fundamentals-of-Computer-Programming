```python
import socket

print("================================")
print("       SERVER BANNER GRABBER")
print("================================")

host = input("Enter server address: ")
port = int(input("Enter port number: "))

try:
    # Create TCP socket
    client = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    client.settimeout(5)

    # Connect to the server
    print("\nConnecting to", host, "on port", port, "...")
    client.connect((host, port))

    print("Connection successful!")

    # Send a simple request
    client.sendall(b"HEAD / HTTP/1.0\r\n\r\n")

    # Receive server response
    banner = client.recv(4096).decode(errors="ignore")

    print("\n--- Server Banner ---")
    print(banner)

    # Close connection
    client.close()

except socket.timeout:
    print("Error: Connection timed out.")

except ConnectionRefusedError:
    print("Error: Connection refused.")

except ValueError:
    print("Error: Please enter a valid port number.")

except Exception as e:
    print("Error:", e)
```
