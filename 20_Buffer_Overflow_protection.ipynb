```python
buffer_size = 10

print("----- Buffer Protection -----")
print("Maximum allowed characters:", buffer_size)

data = input("Enter your data: ")

if len(data) > buffer_size:
    print("\nWarning: Input is too large!")
    print("Allowed characters:", buffer_size)
    print("Entered characters:", len(data))
    print("Data was not stored.")

else:
    print("\nInput is within the limit.")
    print("Data stored successfully.")
    print("Stored data:", data)
```
