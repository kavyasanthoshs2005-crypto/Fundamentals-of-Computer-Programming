```python
import socket

website = input("Enter website name: ")

try:
    ip = socket.gethostbyname(website)
    details = socket.gethostbyname_ex(website)

    print("\n----- Host Details -----")
    print("Website:", website)
    print("IP Address:", ip)
    print("Hostname:", details[0])
    print("Aliases:", details[1])
    print("IP Addresses:", details[2])

except socket.gaierror:
    print("Host could not be found.")
```
