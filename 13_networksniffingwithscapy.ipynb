```python
import random

def show_packet():
    source = "192.168.1." + str(random.randint(2, 20))
    destination = "192.168.1.1"
    protocol = random.choice([6, 17, 1])
    size = random.randint(60, 500)

    print("Source IP      :", source)
    print("Destination IP :", destination)
    print("Protocol       :", protocol)
    print("Packet Size    :", size)
    print("-----------------------------")


print("Capturing 5 packets...")

for i in range(5):
    show_packet()

print("Capture completed.")
```
